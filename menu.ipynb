# **Menu Code**

The widgetsMenu function uses ipywidgets to create a menu for the store manager to navigate through the system with. Each of the clicked_.... functions are used to take the user to the correct screen once a button has been clicked. The btn parameter is used because the on.click() function passes in the button object so the clicked_... function must accept one parameter.

```python
###########################################################################
#Displays a Menu with buttons that allow the users to see different screens
###########################################################################
#Written by F518833 on 2025/11/10

def widgetsMenu():
  #Allows ipywidgets to be used allowing buttons to be used in the menu
  import ipywidgets as widgets
  from ipywidgets import Layout

  #Check subscription button
  def clicked_subs(btn):
    checkSubscription()

  #Search, View and Rent Games button
  def clicked_games(btn):
    games()

  #Return games button
  def clicked_return(btn):
    returnGames()

  #Book in-house session button
  def clicked_book(btn):
    book()

  #Feedback button
  def clicked_feedback(btn):
    feedback()

  #Inventory Pruning button
  def clicked_inv(btn):
    invPrune()

  #Create individual Button widgets
  btnSubs = widgets.Button(description = "Check Subscription", \
                           layout = Layout(width = "50%"))
  btnGames = widgets.Button(description = "Search, View and Rent Games", \
                            layout = Layout(width = "50%"))
  btnReturn = widgets.Button(description = "Return Games", \
                             layout = Layout(width = "50%"))
  btnBook = widgets.Button(description = "Book in-house Session", \
                           layout = Layout(width = "50%"))
  btnFeedback = widgets.Button(description = "Feedback", \
                               layout = Layout(width = "50%"))
  btnInv = widgets.Button(description = "Inventory Pruning", \
                          layout = Layout(width = "50%"))

  #Attach event handlers to buttons
  btnSubs.on_click(clicked_subs)
  btnGames.on_click(clicked_games)
  btnReturn.on_click(clicked_return)
  btnBook.on_click(clicked_book)
  btnFeedback.on_click(clicked_feedback)
  btnInv.on_click(clicked_inv)

  #Display buttons
  display(btnSubs)
  display(btnGames)
  display(btnReturn)
  display(btnBook)
  display(btnFeedback)
  display(btnInv)
```

# **Inventory Pruning**

The invPrune function allows the store manager to select a date which they want to see which games haven't been used since that date. It then gives them the option whether they would then like to remove these games. It uses ipywidgets to allow the user to interact with the system and uses datetime to allow different dates to be compared. The clicked_video function is used to go through the video games file to see which games have been rented since the date entered, the clicked_board function is used to do the same thing with the board games. The clicked_continue function is continue through the system once the date has been entered. The clicked_yes function is used to delete the games which have been flagged to not have been rented since the date. The clicked_no function ensure the user that no games have been removed. The btn parameter is used because the on.click() function passes in the button object so the clicked_... function must accept one parameter.

```python
############################################################
#Allows the user to remove unnecessary games from the system
############################################################
#Written by F518833 on 2026/01/13

def invPrune():
  invList = []
  GameID = ''

  #allows us to compare dates
  from datetime import datetime
  import ipywidgets as widgets

  print("Please enter the date that you would like to check if \
games have been rented after (yyyy-mm-dd)")
  #Create a textbox for the user to input information
  text_box_Date = widgets.Text(
      description='Enter Date:',
      style={'description_width': '150px'}
  )

  #Create individual Button widgets
  btnVideo = widgets.Button(description = "Video")
  btnYes = widgets.Button(description = "Yes")
  btnNo = widgets.Button(description = "No")
  btnBoard = widgets.Button(description = "Board")
  btnContinue = widgets.Button(description = "Continue")

  def clicked_video(btn):
    rented = False
    NotRented = False
    StringDate = text_box_Date.value
    Date = datetime.strptime(StringDate, "%Y-%m-%d").date()
    videoFile = open("Video_Game_Info.txt","r")
    for videoLine in videoFile:
      found = False
      videoValues = videoLine.split(',')
      GameID = videoValues[0]
      rentFile = open("Rental.txt","r")
      for rentLine in rentFile:
        rentValues = rentLine.split(',')
        if rentValues[0] == GameID:
          found = True
          #converts the date from a string to a datetime
          # so that it can be compared
          rentedDate = datetime.strptime(rentValues[1], "%Y-%m-%d").date()
          if Date > rentedDate:
            invList.append(videoLine)
            NotRented = True
          else:
            rented = True

      if found == False:
        invList.append(videoLine)
        NotRented = True

      rentFile.close()
    videoFile.close()

    def clicked_no(btn):
      print("No games have been removed")
      invList = []

    def clicked_yes(btn):
      for i in range(0,len(invList)):
        removeValues = invList[i].split(",")
        videoFile = open('Video_Game_Info.txt','r')
        videoLines = videoFile.readlines()
        videoFile.close()
        videoFile = open("Video_Game_Info.txt","w")
        for videoLine in videoLines:
          videoValues = videoLine.split(",")
          if videoValues[0] != removeValues[0]:
            videoFile.writelines(videoLine)
        videoFile.close()
      print("Games successfully removed")

    if NotRented == False and rented == True:
      print("All games have been rented since the date entered")
    else:
      print("The games to be removed are:")
      for i in range(0,len(invList)):
        values = invList[i].split(",")
        print("%6s | %21s | %15s | %11s | %10s"%(values[0],values[1],values[2],\
                                                 values[3],values[4]))
      print("Would you like to remove these games?")
      display(btnYes, btnNo)
      btnYes.on_click(clicked_yes)
      btnNo.on_click(clicked_no)


  def clicked_board(btn):
    rented = False
    NotRented = False
    StringDate = text_box_Date.value
    Date = datetime.strptime(StringDate, "%Y-%m-%d").date()
    boardFile = open("Board_Game_Info.txt","r")
    for boardLine in boardFile:
      found = False
      boardValues = boardLine.split(',')
      GameID = boardValues[0]
      rentFile = open("Rental.txt","r")
      for rentLine in rentFile:
        rentValues = rentLine.split(',')
        if rentValues[0] == GameID:
          found = True
          rentedDate = datetime.strptime(rentValues[1], "%Y-%m-%d").date()
          if Date > rentedDate:
            invList.append(boardLine)
            NotRented = True
          else:
            rented = True

      if found == False:
        invList.append(boardLine)
        NotRented = True

      rentFile.close()
    boardFile.close()

    def clicked_no(btn):
      print("No games have been removed")
      invList = []

    def clicked_yes(btn):
      for i in range(0,len(invList)):
        removeValues = invList[i].split(",")
        boardFile = open('Board_Game_Info.txt','r')
        boardLines = boardFile.readlines()
        boardFile.close()
        boardFile = open("Board_Game_Info.txt","w")
        for boardLine in boardLines:
          boardValues = boardLine.split(",")
          if boardValues[0] != removeValues[0]:
            boardFile.writelines(boardLine)
        boardFile.close()
      print("Games sucessfully removed")

    if NotRented == False and rented == True:
      print("All games have been rented since the date entered")
    else:
      print("The games to be removed are:")
      for i in range(0,len(invList)):
        values = invList[i].split(",")
        print("%6s | %19s | %15s | %11s | %10s"%(values[0],values[1],values[2],\
                                                 values[3],values[4]))
      print("Would you like to remove these games?")
      display(btnYes, btnNo)
      btnYes.on_click(clicked_yes)
      btnNo.on_click(clicked_no)


  def clicked_continue(btn):
    print("Would you like to complete inventory pruning on board games \
or video games?")
    display(btnVideo, btnBoard)
    btnVideo.on_click(clicked_video)
    btnBoard.on_click(clicked_board)

  display(text_box_Date, btnContinue)
  btnContinue.on_click(clicked_continue)
```

```python
############################################################################
#Checks to see if the game is able to be returned, then returns the game and
#allows the user to leave feedback
############################################################################
#Written by F518833 on 2026/01/03

def returnGames():

  import ipywidgets as widgets

  print("Please enter the GameID of the game being returned")
  #Create a textbox for the user to input information
  text_box_GameID = widgets.Text(
      description='Enter GameID:',
      style={'description_width': '150px'}
  )
  print("Please enter todays date in the format (yyyy-mm-dd)")
  #Create a textbox for the user to input information
  text_box_Date = widgets.Text(
      description='Enter Date:',
      style={'description_width': '150px'}
  )

  #Create a textbox for the user to input information
  text_box_Rating = widgets.Text(
      description='Enter Rating:',
      style={'description_width': '150px'}
  )

  #Create a textbox for the user to input information
  text_box_Comments = widgets.Text(
      description='Enter Comments:',
      style={'description_width': '150px'}
  )

  #Create buttons to allow you to continue the program
  #after the text has been entered
  btnReturn = widgets.Button(description = "Return")
  btnContinue = widgets.Button(description = "Continue")
  btnYes = widgets.Button(description = "Yes")
  btnNo = widgets.Button(description = "No")

  #Once a the return button is clicked the Rental.txt is
  #checked to ensure the game is actually in the file if it
  #is the return date is updated
  def clicked_return(btn):
    found = False
    GameID = text_box_GameID.value
    Date = text_box_Date.value
    file = open('Rental.txt','r')
    lines = file.readlines()
    file.close()
    file = open("Rental.txt","w")
    for line in lines:
      values = line.split(",")
      if values[0] == GameID:
        if values[2] == '':
          found = True
          startDate = values[1]
          customerID = values[3]
          #Replaces the blank space in the file with the new date
          new_line = line.replace(',,',f",{Date},")
          file.write(new_line)
          break
        else:
          file.writelines(line)
      else:
        file.write(line)
    file.close()

    #Once continue is clicked the information is entered into a new line in
    #the game feedback file then tells the user that the return is complete
    def clicked_continue(btn):
        import feedbackManager as fmSL
        GameID = text_box_GameID.value
        Rating = text_box_Rating.value
        Comments = text_box_Comments.value
        fmSL.add_feedback(GameID,Rating,Comments,"Game_Feedback.txt")
        print("Return Complete")

    #If yes is clicked then display the textboxes to allow the
    #necessary data to be entered
    def clicked_yes(btn):
      display(text_box_Rating, text_box_Comments, btnContinue)
      btnContinue.on_click(clicked_continue)

    #If no is clicked then the user doesen't want to give
    #feedback so the return is complete
    def clicked_no(btn):
      print("Return Complete")

    if found == False:
      print("Game not able to be returned")
    elif found == True:
      print("Would you like to give feedback on this game?")
      display(btnYes, btnNo)
      btnYes.on_click(clicked_yes)
      btnNo.on_click(clicked_no)

  display(text_box_GameID, text_box_Date, btnReturn)
  btnReturn.on_click(clicked_return)
```

The checkSubscription function is used to allow the store manager to choose whether they would like to see all customers with a subscription or if they would like to enter a customer id to check if they have subscription.
```python
################################################################
#Displays all the customers with subscription or searches for a
#Specific customer and then checks if they have a subscription
################################################################
#Written by F518833 on 2026/01/07

def checkSubscription():
  print("Would you like to view all customers with a subscription \
or search for a certain customer?")

  import ipywidgets as widgets
  from ipywidgets import Layout

  #View All button
  def clicked_view(btn):
    SubView()

  #Search Customers button
  def clicked_searching(btn):
    SearchCustomers()


  #Create individual Button widgets
  btnView = widgets.Button(description = "View All",\
                           layout = Layout(width = "50%"))
  btnSearch = widgets.Button(description = "Check a specific customer",\
                             layout = Layout(width = "50%"))

  #Attach event handlers to buttons
  btnView.on_click(clicked_view)
  btnSearch.on_click(clicked_searching)

  #Display buttons
  display(btnView)
  display(btnSearch)

  def SubView():
    records = []
    import subscriptionManager as smSL
    subscriptions_list = smSL.load_subscriptions()
    print("CustomerID | Subscription Type |           Start Date |\
             End Date")
    print("-"*77)
    for customer_id, sub in subscriptions_list.items():
      record = {'CustomerID': customer_id}
      record.update(sub)
      #records.append(record)
      value = record
      print("%10s | %17s | %20s | %20s"%(value['CustomerID'],value['SubscriptionType']\
                                         ,value['StartDate'],value['EndDate']))


  def SearchCustomers():
    #Create widget
    text_box_Search = widgets.Text(
      description='Enter CustomerID:',
      placeholder='Type something here ...',
      style={'description_width': '150px'}
    )

    #create a search button
    btnSearch = widgets.Button(description = "Search")

    def clicked_Search(btn):
      found = False
      SearchCustomerID = text_box_Search.value
      records = []
      import subscriptionManager as smSL
      subscriptions_list = smSL.load_subscriptions()
      print("CustomerID | Subscription Type |           Start Date |\
               End Date")
      print("-"*77)
      for customer_id, sub in subscriptions_list.items():
        if customer_id.lower() == SearchCustomerID.lower():
          record = {'CustomerID': customer_id}
          record.update(sub)
          value = record
          print("%10s | %17s | %20s | %20s"%(value['CustomerID'],value['SubscriptionType']\
                                             ,value['StartDate'],value['EndDate']))
          found = True
      if found == False:
        print("Customer doesn't have a subscription")
    display(text_box_Search, btnSearch)
    #Attach event handlers to buttons
    btnSearch.on_click(clicked_Search)

```

# **Book** **In** **-** **House** **Session**

The book function allows users to enter information and then book a session when there is availablity. The clicked_check function checks that the customerID entered has a subscription. The clicked_availablity checks what availability there is and displays the times that are available on the date entered. The clicked_book function allows the data entered to be written to the file and the session is booked. The clicked_2pm and clicked_6pm functions set the time to the correct value when writing to the file. The btn parameter is used because the on.click() function passes in the button object so the clicked_... function must accept one parameter.

```python
##########################################################################
#Allows the user to book an in-house session:
#Customers can book slots 2pm-6pm or 6pm-10pm and can bring up to 3 guests
#There can be 50 people in the store at a time
##########################################################################
#Written by F518833 on 2026/01/10

def book():

  import ipywidgets as widgets
  from ipywidgets import Layout

  text_box_CustomerID = widgets.Text(
      description='Enter CustomerID:',
      style={'description_width': '150px'}
  )
  text_box_Date = widgets.Text(
      description='Enter Date:',
      style={'description_width': '150px'}
  )
  text_box_Guests = widgets.Text(
      description='Enter No. Guests:',
      style={'description_width': '150px'}
  )


  #Create a Button
  btnCheck = widgets.Button(description = "Check CustomerID")
  btnAvailiable = widgets.Button(description = "Check Availablity")
  btnBook = widgets.Button(description = "Book")

  def clicked_check(btn):
      found = False
      CustomerID = text_box_CustomerID.value
      records = []
      import subscriptionManager as smSL
      subscriptions_list = smSL.load_subscriptions()
      for customer_id, sub in subscriptions_list.items():
        if customer_id.lower() == CustomerID.lower():
          record = {'CustomerID': customer_id}
          record.update(sub)
          value = record
          found = True
      if found == True:
        print("Customer has subscription so is eligible to book a session")
        print("Please enter the date for the in-house session")
        display(text_box_Date, btnAvailiable)
        #Attach event handlers to the button
        btnAvailiable.on_click(clicked_availability)
      elif found == False:
        print("Customer does not have a subscription so is \
  unable to book a session")

  def clicked_availability(btn):
    time = ''
    SlotOne = True
    SlotTwo = True
    totalOne = 0
    totalTwo = 0
    Date = text_box_Date.value
    file = open("Bookings.txt","r")
    for line in file:
      value = line.split(",")
      if value[1] == Date:
        if value[2] == '2pm':
          totalOne += 1
        if value[2] == '6pm':
          totalTwo += 1
    file.close()

    if totalOne >= 50:
      SlotOne = False
    if totalTwo >= 50:
      SlotTwo = False

    def clicked_book(btn):
      Guests = text_box_Guests.value
      if int(Guests) > 3:
        print("Too many guests")
      else:
        CustomerID = text_box_CustomerID.value
        file = open("Bookings.txt","a")
        global time
        file.write(f"{CustomerID},{Date},{time},{Guests}\n")
        file.close()
        print("Booking Confirmed")

    def clicked_2pm(btn):
      global time
      time = '2pm'
      display(text_box_Guests, btnBook)
      btnBook.on_click(clicked_book)

    def clicked_6pm(btn):
      global time
      time = '6pm'
      display(text_box_Guests, btnBook)
      btnBook.on_click(clicked_book)

    btn2pm = widgets.Button(description = "2pm - 6pm")
    btn6pm = widgets.Button(description = "6pm - 10pm")
    if SlotOne == True and SlotTwo == True:
      display(btn2pm,btn6pm)
      btn2pm.on_click(clicked_2pm)
      btn6pm.on_click(clicked_6pm)
    elif SlotOne == True:
      display(btn2pm)
      btn2pm.on_click(clicked_2pm)
    elif SlotTwo == True:
      display(btn6pm)
      btn6pm.on_click(clicked_6pm)


  display(text_box_CustomerID, btnCheck)
  #Attach event handlers to the button
  btnCheck.on_click(clicked_check)
```
